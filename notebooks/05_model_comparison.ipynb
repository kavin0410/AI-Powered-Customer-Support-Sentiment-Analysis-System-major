### 3. End-to-End Real-Time Inference Testing


```python
sample_queries = [
    'The delivery driver was two days late and crushed the package box.',
    'I received a duplicate charge on my credit card statement for order #40291.',
    'Sarah from customer support resolved my issue within five minutes! Best service ever.',
    'The mobile app throws error 500 every time I open the checkout screen.',
    'Please consider adding eco-friendly packaging options in your future store updates.'
]

results = []
for text in sample_queries:
    pred = predict_feedback(text)
    results.append({
        'Feedback': text,
        'Sentiment': pred['sentiment'],
        'Sentiment Confidence': f"{pred['sentiment_confidence']:.2%}",
        'Issue Category': pred['issue_category'],
        'Issue Confidence': f"{pred['issue_confidence']:.2%}"
    })

pd.DataFrame(results)
```
