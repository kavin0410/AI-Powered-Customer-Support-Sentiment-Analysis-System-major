# 01. Data Collection & Raw Dataset Inspection
**Project**: AI-Powered Customer Support & Sentiment Analysis System

This notebook covers:
- Overview of data schema and target classes
- Reproducible dataset generation pipeline for customer support feedback
- Injection of realistic data anomalies (duplicates, missing text, irregular casing)
- Raw dataset validation and initial inspection


```python
import os
import sys
sys.path.insert(0, '..')

import pandas as pd
from src.data_generator import generate_raw_dataset, CATEGORIES, SENTIMENTS

print('Categories:', CATEGORIES)
print('Sentiments:', SENTIMENTS)
```

### 1. Generate or Load Raw Feedback Dataset


```python
raw_csv_path = '../data/raw/raw_feedback.csv'
if not os.path.exists(raw_csv_path):
    df_raw = generate_raw_dataset(n_samples=3200, random_seed=42)
    os.makedirs('../data/raw', exist_ok=True)
    df_raw.to_csv(raw_csv_path, index=False)
else:
    df_raw = pd.read_csv(raw_csv_path)

print(f'Raw dataset shape: {df_raw.shape}')
df_raw.head()
```

### 2. Inspect Missing Values and Duplicates in Raw Data


```python
print('Missing Values in Raw Data:')
print(df_raw.isnull().sum())

duplicate_count = df_raw.duplicated(subset=['feedback_text', 'sentiment', 'issue_category']).sum()
print(f'\nNumber of duplicate records: {duplicate_count}')
```

### 3. Raw Class Distributions


```python
print('Raw Sentiment Distribution:')
print(df_raw['sentiment'].str.strip().str.capitalize().value_counts())

print('\nRaw Issue Category Distribution:')
print(df_raw['issue_category'].str.strip().str.title().value_counts())
```
